```python
import numpy as np

analyze_mano = True

if analyze_mano:
    data = np.load('/home/erik/msthesis/data/tests/mano_vae_test.npy', allow_pickle=True)
else:
    data = np.load('/home/erik/msthesis/data/tests/gc_angles_vae_test.npy', allow_pickle=True)

gt_data = data[0]
vae_data = data[1]

if not analyze_mano:
    gt_data = gt_data * np.pi / 180

print(f'gt_data shape: {gt_data.shape}')
print(f'vae_data shape: {vae_data.shape}')

```

```python
import matplotlib.pyplot as plt

# Number of dimensions
n_dims = min(gt_data.shape[1], 15)

# Create a figure with subplots for each dimension
fig, axes = plt.subplots(n_dims, 1, figsize=(12, 4*n_dims), sharex=True)
fig.suptitle('Ground Truth vs VAE Output for Each Dimension', fontsize=16)

# Time axis
time = np.arange(gt_data.shape[0])

# Plot each dimension
for i in range(n_dims):
    axes[i].plot(time, gt_data[:, i], color='green', label='Ground Truth', alpha=0.7)
    axes[i].plot(time, vae_data[:, i], color='red', label='VAE Output', alpha=0.7)
    axes[i].set_ylabel(f'Dimension {i+1}')
    axes[i].legend()
    axes[i].grid(True)

# Set common x-label
axes[-1].set_xlabel('Time')

plt.tight_layout()
plt.show()
```
